# 1. Calcul d'un prix

Ecrivez un programme faisant saisir à l’utilisateur un prix unitaire HT, un taux de TVA et un nombre d’articles, et qui calcule le montant TTC de l’achat.<br>
Affichez le montant TTC.

```python
prix_unitaire_ht = float(input("Entrez votre prix unitaire HT : "))    # Nombre décimal
taux_tva = float(input("Entrez votre taux de TVA en % : "))            # Nombre décimal
nb_article = int(input("Entrez votre nombre d'article : "))            # Nombre entier 

montant_ht = prix_unitaire_ht * nb_article                             # Calcul du montant HT
montant_ttc = montant_ht * (1 + taux_tva / 100)                        # Calcul du montant TTC avec HT

print("Votre montant TTC est de :", montant_ttc)
```

# 2. Conversion de durées (1)

Ecrivez un programme demandant une durée en secondes à l’utilisateur, et qui la convertit en heures, minutes, secondes.<br>
Affichez le résultat.<br>
Par exemple 12334 deviendra 3 heures, 25 minutes et 34 secondes.

```python
duree_s = int(input("Entrez un nombre de secondes : "))
heures = duree_s // 3600
heures_reste = duree_s % 3600
minutes = heures_reste // 60
secondes = heures_reste % 60
print("Votre durée (h/m/s) est la suivante : ", heures, "heures", minutes, "minutes", secondes, "secondes")
```

# 3. Conversion de durées (2)

Ecrivez un programme demandant une durée en heures, minutes, secondes à l’utilisateur et qui la convertit en secondes.<br>
Affichez le résultat.<br>
Par exemple 3 heures, 25 minutes et 34 secondes deviendra 12334 secondes.

```python
heures = int(input("Entrez un nombre d'heures : "))
minutes = int(input("Entrez un nombre de minutes : "))
secondes = int(input("Entrez un nombre de secondes : "))
secondes_total = (heures * 3600) + (minutes * 60) + secondes
print("Voici votre durée convertie en secondes : ", secondes_total)
```
