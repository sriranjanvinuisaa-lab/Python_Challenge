# Mission 1

What's the lenght of the following string ?

````python
s = "Taumata whakatangihanga koauau o tamatea turi pukaka piki maungah oronuku pokai whenuaki tanatahu "
print(len(s)) # Donc : 98 carctères avec espaces y compris
````

Now, count only non-space characters and display how many there is.
Remember, using a for loop on a string will browse all the individual characters of it!

````python
s = "Taumata whakatangihanga koauau o tamatea turi pukaka piki maungah oronuku pokai whenuaki tanatahu "
compteur = 0                # Initialisation du compteur à 0
for caractere in s:         # Remember, using a for loop on a string will browse all the individual characters of it!, donc pas besoin len()
    if caractere != " ":    # Si seulement le caractère n'est pas une espace " " 
        compteur += 1       # Nombre de caractère sans espace
print(compteur)             # Donc : 85 caractères sans espaces
````

# Mission 2

Ask the user for a string, then for a character (we assume user input follows the rules). Then, display the positions, **starting with 1**, of all occurrences of the character within the string.

For example, for the word "welcome" and for the character "e", the script will display:
```
position: 2
position: 7
```

````python
string = input("Enter a string: ")
caractere = input("Enter a character: ")
pos = 1                                      # Initialisation à 1 pour le compter le nombre de position, dans le string, il doit y'avoir au moins un caractère

for lettre in string:                        # la variable "lettre" parcourt le string par caractère
    if lettre == caractere:                  # Si "lettre" est égale au caractère 
        print("position:", pos)              # alors on affiche la position du caractère
    pos += 1                                 # Nombre de positions 
````

# Mission 5

Ask the user for a string, then for a character, then for another character.

Then, display the string where the occurences of the first character are replaced by the second one.  
For example:
* string = `"Boulgour"`  
* 1st_char = `"g"`
* 2nd_char = `"v"`

Result : `"Boulvour"`

````python
string = input("Enter a string: ")
char_1 = input("Enter a character: ")
char_2 = input("Enter a character: ")
result = string.replace(char_1, char_2)
print(result)
````

# Mission 6

Ask the user for a string. Then, display the same string with all vowels converted to uppercase, and all consonants converted to lowercase.  
For example:
* string = `"Papa's got a brand new bag"`  
Result : `"pApA's gOt A brAnd nEw bAg"`


````python
string = input("Enter a string: ")
resultat = ""                        
for lettre in string:                    
    if lettre.lower() in "aeiou":     # Je met tout en minuscule au cas où il y a également les voyelles en majuscule          
        resultat += lettre.upper()    # Pour les voyelles, en majuscule
    else:
        resultat += lettre.lower()    # Pour les consonnes, reste en minuscule
print(resultat)
````
