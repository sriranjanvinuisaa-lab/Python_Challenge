# Mission 1

For each cell, execute the code and answer the question before moving on to the next.

```python
names = ["Fredy", "Nicolas", "Alexandre"]

for i in enumerate(names):   # On voit qu'il affiche la position de chaque mot (index) et le mot associé
  print(i)
  print(type(i))
```

1. What type of object is returned as output?  
Hint: You can use `type(i)`.

Answer: Le type de l'objet est un tuple. 

```python
for i, val in enumerate(names):
  print(i, val)
```

2. What does `i` correspond to?  
What does `val` correspond to?  
Why is it so?

Answer: 
"i" correspond à l'index de chaque valeur que contient la liste "names".
"val" correspond à la valeur qui correspond à l'index i. 
Grâce à la commande "enumerate(names)", il parcours la liste pour indiquer la position (i) de chaque valeurs (val). 

```python
for val, i in enumerate(names):
  print(val)
```

3. This time, `i` returns first names and `val` returns numbers. Why is it so?

Answer: 
Python attribue toujours : 
La première valeur à la première variable donc ici (val).
La deuxième valeur à la deuxième variable donc ici (i).

4. What does the built in function `enumerate` do?
In what cases can it be useful?

Answer: la fonction "enumerate" sert à indiquer la position et la valeur de chaque élément dans une liste. 
Il est utile dans le cas où on veut connaitre la position de chaque valeur dans une liste.

1. What can the `.join()` method do?  
What is the type of `test_list` before using `.join()`?  
What is the type of `test_list` after using `.join()`?

Answer:
La méthode "join()" permet de regouper tout les élément en une seul chaine de caractères. 
Avant d'utiliser join(), test_list est une liste "list()" composés de plusieurs éléments. 
Après avoir utliser join(), test_list est une chaîne de caractère "str()" contenant tout les éléments séparés par un espace. 

```python
christmas = "Noël c'est bien, mais attention à ne pas manger trop de chocolat. On le sait, mais on le fais à chaque fois. C'est la vie"
christmas.split()
```

2. What can the `.split()` method do?  
What is the type of `christmas` before using `.split()`?  
What is the type of `christmas` after using `.split()`?

Answer: La méthode split(), sépare une chaine de caractères en plusieurs éléments en utilisant les espaces comme séparateurs. 
Avant l'utilisation de split(), elle est de type str(), chaine de caractères. 
Après l'utilisatin de split(), elle est une liste (list()) contenant les éléments séparés.

```python
christmas = "Noël c'est bien, mais attention à ne pas manger trop de chocolat. On le sait, mais on le fait à chaque fois. C'est la vie"

christmas.split("à")
```

3. Why does the `.split()` method split the `string` into 3 elements?

Answer:
La méthode split("à") sépare le texte à chaque fois qu'elle rencontre le caractère "à". 
Dans la chaine de caractères, "à" apparaît deux fois donc le texte se sépare en 3 éléments.
