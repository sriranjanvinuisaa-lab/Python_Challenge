# Mission 1

Define a list, then display only the elements (and their position) that have even positions (0, 2, 4, 6...).

With `test_list = ["a", "b", "c", "d", "e"]`, the script should display:
```
a at position 0
c at position 2
e at position 4
```

````python
test_list = ["a", "b", "c", "d", "e"]

for i, val in enumerate(test_list):              # i récupère l'index, val récupère les éléments 
    if i % 2 == 0:                               # Vérifie si i, l'index est pair (divisible par 2) 
        print(val, "au position", i)
````

# Mission 4

Define a list, then ask the user for a string. Then, insert the string in the list at a random position, and display the resulting list.

For example, with the list `test_list4 = ["p", "y", "t", "h", "o"]` and the input `n`, you *might* get:
```
["p", "n", "y", "t", "h", "o"]
```
... but you might also get:
```
["p", "y", "t", "n", "h", "o"]
```


````python
import random

test_list4 = ["p", "y", "t", "h", "o"]
string = input("Enter a string : ")

index = random.randint(0, len(test_list4))  # len() donne la longueur de la liste 
test_list4.insert(index, string)

print(test_list4)
````

# Mission 6

Ask the user for a string, convert it to a list of characters with the `list` function, then ask for a single character.

Remove the first occurrence of the character from the list, then display how many occurrences of the character remain in the list.

For instance if you enter `"Excellent work"` then the letter `"e"`, it will display:
```
There are still 2 copies of e in the list
```

````python
# Juste un petit rappel pour moi : Python différencie les caractères minuscules et majuscules 
# Exemple : e et E sont différents

string = input("Enter a string : ").lower()
liste_string = list(string)
single = input("Enter a single character : ").lower()

liste_string.remove(single)
print("There are still", liste_string.count(single), "copies of", single, "in the list")
````

# Mission 7

Define three lists containing no duplicates. Then, indicate (however you like) the two that have the most elements in common (or all three if there's a tie).

For instance, with:
```
list_a = [1, 2, 4, 8, 16, 32]
list_b = [1, 2, 3, 5, 8, 13]
list_c = [2, 3, 5, 7, 11, 13]
```
It may display:
```
Lists B and C have the most common elements!
```

````python
list_a = [1, 2, 4, 8, 16, 32]
list_b = [1, 2, 3, 5, 8, 13]
list_c = [2, 3, 5, 7, 11, 13]

list_common_ab = 0               # Initialisation à 0
for i in list_a:                 # Pour i dans la liste a 
    if i in list_b:              # Si i apparait dans la liste b
        list_common_ab += 1         # alors la variable common_ab compte le nombre d'éléments (i) commun

list_common_ac = 0
for i in list_a:
    if i in list_c:
        list_common_ac += 1

list_common_bc = 0
for i in list_b:
    if i in list_c:
        list_common_bc += 1

## Display
        
if list_common_ab >= list_common_ac and list_common_ab >= list_common_bc:
    print("Lists A and B have the most common elements!")

if list_common_ac >= list_common_ab and list_common_ac >= list_common_bc:
    print("Lists A and C have the most common elements!")

if list_common_bc >= list_common_ab and list_common_bc >= list_common_ac:
    print("Lists B and C have the most common elements!")
````
