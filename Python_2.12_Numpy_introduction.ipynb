```python
# Mission 3 : Create a function that uses the np.random.random() function to return an array with 10 random elements,
# all of which must be greater than 0.25 and less than 1.0

# Example of expected result: array([0.3897948 , 0.82850041, 0.5499838 , 0.81940137, 0.92175875, 0.89634373, 0.26821861, 0.72376496, 0.81098623, 0.28192741])

def random_array():
    """
    Paramètre : aucun.
    Retourne 10 nombres aléatoires strictement supérieurs à 0.25
    et strictement inférieurs à 1.
    np.random.random() génère des nombres dans [0, 1) :
    0 est inclus et 1 est exclu.
    """
    aleatoire = np.random.random(10)     # Génère 10 nombres entre 0 et 1

    while np.min(aleatoire) <= 0.25:        # Tant que le minimum reste inférieur ou égal à 0.25,
        aleatoire = np.random.random(10)    # on génère un nouveau tableau de 10 nombres.

    return aleatoire                           # Retourne un tableau qui respecte les conditions

print(random_array())
```
