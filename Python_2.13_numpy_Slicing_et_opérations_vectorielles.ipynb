# Mission 2

Modify, using a `numpy` method, the `array_1` object to be an object that has two axes, including 1 row and 32 columns.  
Save it under the variable `array_2`.

```python
array_2 = array_1.reshape(1, 32)    # 1 ligne et 32 colonnes
array_2
```

Now, modify `array_1` into an object that has 32 rows and only 1 column.  
Save it under the variable `array_3`.

```python
array_3 = array_1.reshape(32, 1)  # 32 lignes et 1 colonne
array_3
```

Access the first element of `array_2`, then of `array_3`.  
What do you notice?  
Why ?

```python
print(array_2[0])    # 0 : Indice de la ligne 1
print(array_3[0])    # 0 : Indice de la ligne 1

# En utilisant la fonction reshape(), nous avons transformé les tableaux en tableaux à 2 dimensions.
# array_2 est composé d'une ligne et de 32 colonnes. L'indice 0 sélectionne toute la première ligne, donc les 32 éléments. 
# array_3 est composé de 32 lignes et d'une colonne. L'indice 0 sélectionne toute la première ligne, donc un seul élément.

# J'ai pensé au premier élément du tableau, donc :
print(array_2[0, 0])    # 0 : Indice de la ligne et de la colonne 1
print(array_3[0, 0])    # 0 : Indice de la ligne 1 et de la colonne 
```

# Mission 4

Reshape `array_1` to be a 4-row, 8-column array,
rather than a 32-element vector.  
Save your array under the variable `array_4`.

```python
array_4 = array_1.reshape(4, 8)     # 4 lignes et 8 colonnes
array_4
```

Access for each row, the values ​​of the columns which have the index 1 to 2 included.

```python
print(array_4[:, 1:3])

# Toutes les lignes avec ":"
# Les colonnes d'indices 1 et 2 avec "1:3".
# L'index 3 exclu.
```

Access for rows at position 1 and 2 inclusive, the values ​​of all columns.

```python
print(array_4[0:2, :])

# La ligne position 1 : indice 0 et la ligne position 2 : indice 1.
# Toutes les colonnes avec ":"
```

Access for the rows which have the index 2 to 3 inclusive, to the values ​​of the columns which have the index 3 to 5 inclusive.

```python
array_4[2:4, 3:6]

# Lignes : index 2 à 3 inclus, 4 exclu
# Colonnes : index 3 à 5 inclus, 6 exclu
```

Select the first column, as well as the columns whose index is a multiple of 2.

```python
print(array_4[:, ::2])

# Les indices des colonnes commencent à 0.
# "::2" sélectionne une colonne tous les 2 indices.
# 0, 2, 4, 6 (indices pairs).
```

Select columns whose index is odd.

```python
print(array_4[:, 1::2])

# En commençant à l'indice 1, "1::2" sélectionne une colonne tous les 2 indices.
# 1, 3, 5, 7 (indices impairs).
```

Access row values ​​at positions 2 to 3,

```python
print(array_5[1, 1:3, 1:])

## selection de la matrice 2 (index 1)
# selection des lignes aux positions 2 et 3 (index 1 et index 2). L'index 3 est exclu.
# selection des colonnes aux index 1 à 3. 
```
